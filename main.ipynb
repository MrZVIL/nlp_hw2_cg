```python
import os
import wandb
from kaggle_secrets import UserSecretsClient
from huggingface_hub import HfApi
from datasets import load_dataset
import torch
import numpy as np
import random
```

```python
secrets = UserSecretsClient()

wandb.login(key=secrets.get_secret("wandb"))

hf_api = HfApi(token=secrets.get_secret("huggingface-cli"))
print("HF account:", hf_api.whoami()["name"])
```

```python
os.environ['HF_HOME'] = '/kaggle/working/hf_hub'
```

```python
experiment_name = "logging-smoke-test"

run = wandb.init(
    project="hw2-code-generation",
    name=experiment_name,
    config={"purpose": "check_connection"},
)

run.log({"test/value": 1}, step=0)
run.finish()
```

```python
def set_global_seed(seed: int) -> None:
    """Set global seed for reproducibility.
    :param int seed: Seed to be set
    """
    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
SEED = 67
set_global_seed(SEED)
```

```python
hf_user_name = hf_api.whoami()["name"]
repo_id = f"{hf_user_name}/hw2-cg-{experiment_name}"

hf_api.create_repo(
    repo_id=repo_id,
    repo_type="model",
    private=True,
    exist_ok=True,
)
```

```python
dataset = load_dataset("SlavaYus/tinysms_qa_12m")
print(dataset)
```

```python
split = dataset["train"].train_test_split(
    test_size=0.10,
    shuffle=True,
    seed=SEED
)

train_dataset = split["train"]
valid_dataset = split["test"]
test_dataset = dataset["test"]

print("Train:", len(train_dataset))
print("Valid:", len(valid_dataset))
print("Test:", len(test_dataset))
```
